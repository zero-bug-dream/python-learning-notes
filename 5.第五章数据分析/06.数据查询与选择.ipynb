```python
# 数据选择
df1 = pd.read_csv('data/sales_data_500.csv',usecols = ['订单号','产品类别','产品名称','销售数量','单价'])
df1
# 选择列
# 1.1 单列
df1['产品类别']
# 1.2 多列
df1[['产品名称','单价']]

# 选择行
# 2.1 单行 -语法: df.iloc[start:stop:step] (基于行号选择,从0开始)
df1.iloc[0:5:1]
# 2.2 多行 -语法: df.loc[start:stop:step] 包含结束位置 (只看标签选择)
df1.loc[0:5:1]


```

```python
df2 = pd.read_csv('data/sales_data_500.csv')
# 数据过滤

# 销售数量>=10 df[filter条件]
df2[df2['销售数量'] >= 10]

# 产品类别是食品或图书的订单
# df2[(df2['产品类别'] == '食品') | (df2['产品类别'] == '图书' )]
df2[df2['产品类别'].isin(['食品','图书'])]

# 单价在100 - 200间 (between:包含100和200)
df2[(df2['单价'].between(100,200))]

# 销售数量>=8且单价>=100的订单 (&:逻辑与, |:逻辑或)
df2[(df2['销售数量'] >= 8) & (df2['单价'] >= 100)]

# 产品类别为服装或食品,支付方式为支付宝或微信支付
df2[df2['产品类别'].isin (['服装','食品']) & df2['支付方式'].isin (['支付宝','微信支付'])]
```
