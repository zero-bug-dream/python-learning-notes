```python
import pandas as pd
df = pd.read_csv('data/sales_data_500.csv',nrows = 20)

# 1.0排序 df.sort_values(by = '销售数量',ascending = False)
# 1.1根据 销售数量 降序排序
# df.sort_values(by = '销售数量',ascending = False)
# 1.2根据 价格 升序排序
df.sort_values(by = '单价',ascending = True)
# 1.3根据 价格 升序排序,价格一样,再根据 销售数量 排序(列表的形式传递)
df.sort_values(by = ['单价','销售数量'],ascending = [True,False])
```

```python
df = pd.read_csv('data/sales_data_500.csv',nrows = 20)
df['销售金额'] = df['单价'] * df['销售数量']
# 3.0 分组
# 3.1 根据 产品类别 分组,统计各类别的 订单数量
df.groupby('产品类别').count()
# 3.2 根据 产品类别 分组,统计各类别的 销售数量之和
df.groupby('产品类别')['销售数量'].sum()
# 3.3 根据 产品类别 分组,统计各类别的 销售金额之和
df.groupby('产品类别')['销售金额'].sum()
# 3.4 根据 产品类别 分组,统计各类别的 最小单价
df.groupby('产品类别')['单价'].min()
# 3.5 根据 产品类别 分组,统计各类别的 最大单价
df.groupby('产品类别')['单价'].max()
# 3.6 根据 产品类别 分组,统计各类别的 平均单价
df.groupby('产品类别')['单价'].mean()
# 3.7 根据 产品类别 分组,统计各类别的 单价的平均值、最大值、最小值
df.groupby('产品类别')['单价'].agg(['mean','max','min']).round(2)
# 3.8 根据 产品类别 分组,统计各类别的 销售数量之和、销售金额之和、单价的平均值
df.groupby('产品类别').agg({'销售数量':'sum','销售金额':'sum','单价':['mean']}).round(2).to_dict()



```
