```python
import pandas as pd
# 构造DataFrame --- 创建数据集(学员成绩) 列表中字典
a = pd.DataFrame([
    {"姓名": "张三", "数学": 80, "英语": 90, "物理": 75},
    {"姓名": "李四", "数学": 75, "英语": 85, "物理": 80},
    {"姓名": "王五", "数学": 90, "英语": 88, "物理": 92},
    {"姓名": "赵六", "数学": 85, "英语": 95, "物理": 88},
    {"姓名": "小王1", "数学": 78, "英语": 82, "物理": 88},
    {"姓名": "小王2", "数学": 88, "英语": 92, "物理": 85},
    {"姓名":"小王3", "数学": 90, "英语": 85, "物理": 90}
])


# 统计计算(series列表的一列)

print(f"数学的最高分:{a['数学'].max()},最低分:{a['数学'].min()},平均分:{a['数学'].mean():.2f}")
print(f"英语的最高分:{a['英语'].max()},最低分:{a['英语'].min()},平均分:{a['英语'].mean():.2f}")

print(f"物理的最高分:{a['物理'].max()},最低分:{a['物理'].min()},平均分:{a['物理'].mean():.2f}")
```
