```python
# 数据清洗:处理缺失值,重复值,异常值
import pandas as pd
pd.set_option('display.min_rows', 30)
# 读取数据
df = pd.read_csv('data/sales_data_500.csv')

df.head()

# 2.数据清洗
# 2.1缺失值处理
df.isnull() # 查看缺失值

# 2.1.1删除缺失值所在行
# df.dropna()
# df = df.dropna(axis=1) # 删除缺失值所在列 0:行 1:列

# 2.1.2填充缺失值
df.fillna("--")
df.ffill() # 填充上一行的值
df.bfill() # 填充下一行的值



```

```python
# 2.2重复值处理
# 2.2.1查看重复值
df.duplicated() # 所有列重复都重复
df.duplicated(subset=['订单号']) #查看指定列重复

# 2.2.2删除重复值
df.drop_duplicates(subset = ['订单号']) #keep = 'first'保留第一个 'last'最后一个
```

```python
# 2.3异常值处理
# 2.3.1查看异常值
df[df["单价"]<0]
df[df["销售数量"]<0]

# 2.3.2删除异常值
df.drop(df[df["销售数量"]<0].index)
df.drop(df[df["单价"]<0].index)

# 2.3.3 修复异常值
df['单价'] = df['单价'].abs() # 取绝对值
df
```

```python
# 2.4 数据格式处理
df['订单日期'] = df['订单日期'].str.replace('.', '/')
df

```
