```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei'] # 显示字体为黑体

# 创建子图(subplots方法):一个画布(figure)包含多个子图(axes) 2行2列(0开始)
figure,axes = plt.subplots(nrows=2, ncols=2,figsize=(30,15),dpi=100) # axes:子图数组    dpi:像素

# 柱状图(axes[1,1])-电力资源储备---bar方法
countries = ['美国', '中国', '印度', '日本', '德国', '法国', '英国', '俄罗斯', '意大利', '加拿大'] # 10个国家列表
powers = [1900, 2700, 500, 300, 200, 150, 100, 800, 120, 600]
axes1:Axes = axes[1,1]
axes1.bar(countries, powers, color='r', width=0.7)
axes1.set_title('电力资源储备',fontsize = 20)
axes1.set_xlabel('国家',fontsize = 15)
axes1.set_ylabel('电力资源储备(千千瓦时)',fontsize = 15)
axes1.grid(linestyle='--',alpha = 0.3)

# 饼状图(世界人口比例)---擅长比例分析---pie方法
countries1 = ['中国','美国','新加坡','加拿大','巴西','印度','俄罗斯','其他']
populations = [1400, 330, 500, 370, 210, 1300, 200, 100]
axes2:Axes = axes[0,0]
axes2.pie(populations, labels=countries1,autopct='%1.2f%%', startangle=90, radius=1.2)
axes2.set_title('世界人口比例',fontsize = 20)
axes2.legend(loc='lower center',ncol=4,bbox_to_anchor=(0.5,-0.2)) # bbox_to_anchor=(0.5,-0.2)):0.5水平居中，-0.2竖直方位

# 保存图片
figure.savefig('data/01.png')

```
