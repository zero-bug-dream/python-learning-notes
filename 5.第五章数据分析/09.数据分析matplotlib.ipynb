```python
import matplotlib.pyplot as plt
import random
# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei'] # 显示字体为黑体


# 绘制折线图
x = [i for i in range(1,25)]
y_js = [random.randint(10,20) for i in x]
y_xj = [random.randint(10,20) for i in x]

plt.figure(figsize=(10,5)) #画布大小 (宽,高)
plt.plot(x, y_xj,label = '新疆') # 绘制折线图 (没有画布(figure),自动会创建
plt.plot(x, y_js,label = '江苏') # 绘制折线图 (没有画布(figure),自动会创建

plt.title('气温变化折线图', fontsize=16) # 标题
plt.xlabel('时间', fontsize=14) # X轴标签
plt.ylabel('温度', fontsize=14) # Y轴标签
# plt.xticks(x[1::2])
plt.xticks(x)
y_ticks = [i for i in range(5,21)]

plt.grid(linestyle='-.',alpha = 0.7)# alpha:透明度
plt.legend(loc = 'upper right')

plt.yticks(y_ticks)
plt.show()  # 显示图标
```
